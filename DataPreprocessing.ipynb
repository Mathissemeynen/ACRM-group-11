```python
# === Convert filename (YYYYDOY) → datetime ===
def doy_to_date(filename: Path):
    return pd.to_datetime(filename.stem, format="%Y%j")
```

```python
def preprocess_with_time(ds):
    """Preprocessing die tijd direct uit filename haalt"""

    # 1. Doe je normale preprocessing
    var_name = list(ds.data_vars)[0]
    da = ds[var_name]

    if "band" in da.dims:
        da = da.isel(band=0)

    da = da.rename({"x": "lon", "y": "lat"})
    da = da.where(da != -9999)
    da = da / 10.0

    # 2. Haal tijd uit filename en voeg toe als coordinaat
    filename = ds.encoding["source"]
    date = doy_to_date(Path(filename))  # Gebruik je bestaande functie!

    # Voeg tijd toe als coordinaat
    ds_processed = da.to_dataset(name="china_temp")
    ds_processed = ds_processed.expand_dims(time=[date])  # Voeg tijd dimensie toe

    return ds_processed
```

## Processing of Global (daily temperature) data

```python
# === combine datasets ===
ds_global_2016 = xr.open_mfdataset(
    global_2016_files,
    combine="by_coords",      # uses time coordinate already stored in each file
    engine="h5netcdf",        # efficient engine for .nc files
    parallel=True,            # enables parallel I/O
    chunks={"lat": 256, "lon": 256, "time": 1},  # safe chunking
)
```

```python
# Rename main variable to "global_temp" --> consistency
var_name = list(ds_global_2016.data_vars)[0]  # likely "air_temperature"
ds_global_2016 = ds_global_2016.rename({var_name: "global_temp"})

ds_global_2016.attrs.update({
    "description": "Global daily air temperature (low resolution)",
    "units": "°C",
    "year": 2016,
})
```

## Processing of China Data

```python
# === time coordinates ===
dates = [doy_to_date(f) for f in china_files_2016]
```

```python
# Nu hoef je geen aparte tijd toe te voegen!
ds_china_2016 = xr.open_mfdataset(
    china_files_2016,
    combine="nested",
    concat_dim="time",
    preprocess=preprocess_with_time,  # Nu zit tijd er al in!
    engine="rasterio",
    chunks={"x": 512, "y": 512},
)
# GEEN ds_china_2016.assign_coords(time=dates) meer nodig!
```

```python
print(ds_china_2016)
```

## Matching


```python
# === temporal matching ===
china_times = np.array(ds_china_2016.time.values)
global_times = np.array(ds_global_2016.time.values)

# Find intersection of timestamps
common_times = np.intersect1d(global_times, china_times)

print(f"Common dates: {len(common_times)} days overlap")

# Subset both datasets to those dates
ds_global_matched = ds_global_2016.sel(time=common_times)
ds_china_matched  = ds_china_2016.sel(time=common_times)
```

```python
# === spatial matching ===
china_mask_hr = ds_china_matched["china_temp"].notnull().any(dim="time")
china_mask_hr_num = china_mask_hr.astype(float)
china_mask_on_global_num = china_mask_hr_num.interp(
    lat=ds_global_matched.lat,
    lon=ds_global_matched.lon,
    method="nearest",
)

china_mask_on_global = china_mask_on_global_num > 0.5
ds_global_china = ds_global_matched.where(china_mask_on_global)
ds_china_china = ds_china_matched.where(~ds_china_matched["china_temp"].isnull())
```


## Train and test (time split)

```python
# === Train/Test split op basis van tijd ===
split_date = "2016-09-01"  # bijvoorbeeld: alles vóór september = train


ds_global_train = ds_global_china.sel(time=slice(None, split_date))
ds_china_train  = ds_china_matched.sel(time=slice(None, split_date))

ds_global_test = ds_global_china.sel(time=slice(split_date, None))
ds_china_test  = ds_china_matched.sel(time=slice(split_date, None))

print("Train periode:", str(ds_global_train.time[0].values)[:10], "→", str(ds_global_train.time[-1].values)[:10])
print("Test  periode:", str(ds_global_test.time[0].values)[:10], "→", str(ds_global_test.time[-1].values)[:10])
print("Train dagen:", ds_global_train.sizes["time"])
print("Test  dagen:", ds_global_test.sizes["time"])
```

```python
print("gebruik eerste paar pixels")

# Sample een paar pixels voor snelle check
sample_lat, sample_lon = 1000, 1000  # Midden in het gebied
train_sample = ds_china_train.china_temp.isel(
    lat=sample_lat, lon=sample_lon, time=slice(0, 10)
).compute()
test_sample = ds_china_test.china_temp.isel(
    lat=sample_lat, lon=sample_lon, time=slice(0, 10)
).compute()

print(f"Train sample: {train_sample.mean().values:.1f}°C")
print(f"Test sample:  {test_sample.mean().values:.1f}°C")
print(f"Verschil: {abs(train_sample.mean() - test_sample.mean()).values:.1f}°C")
```
